```python
from pathlib import Path
import pandas as pd
```

```python
DATA_DIR = Path("data")

classes = [
    "protocol",
    "informed_consent",
    "statistical_analysis_plan",
    "safety_report"
]
```

```python
for cls in classes:
    folder = DATA_DIR / cls
    print(cls, "->", folder.exists())
```

```python
rows = []

for cls in classes:
    folder = DATA_DIR / cls
    files = list(folder.glob("*"))
    
    rows.append({
        "class": cls,
        "file_count": len(files),
        "files": [f.name for f in files]
    })

df = pd.DataFrame(rows)
df[["class", "file_count"]]
```

```python
for cls in classes:
    print(f"\n===== {cls.upper()} =====")
    folder = DATA_DIR / cls
    
    for file in folder.glob("*"):
        print(file.name)
```

```python
ext_rows = []

for cls in classes:
    folder = DATA_DIR / cls
    
    for file in folder.glob("*"):
        ext_rows.append({
            "class": cls,
            "file_name": file.name,
            "extension": file.suffix.lower(),
            "size_mb": round(file.stat().st_size / (1024 * 1024), 2)
        })

files_df = pd.DataFrame(ext_rows)
files_df.head()
```

```python
files_df = files_df[
    ~files_df["file_name"].str.startswith(".")
].reset_index(drop=True)

files_df.head()
```

```python
files_df.groupby("class").size().reset_index(name="count")
```

```python
files_df.groupby(
    ["class", "extension"]
).size().reset_index(name="count")
```

```python
allowed_extensions = [".pdf", ".doc", ".docx", ".txt"]

unsupported = files_df[
    ~files_df["extension"].isin(allowed_extensions)
]

unsupported
```

```python
summary = files_df.groupby("class").agg(
    total_files=("file_name", "count"),
    avg_size_mb=("size_mb", "mean"),
    min_size_mb=("size_mb", "min"),
    max_size_mb=("size_mb", "max")
)

summary
```

```python
files_df.to_csv(
    "dataset_inventory.csv",
    index=False
)

print("Inventory saved successfully.")
```
